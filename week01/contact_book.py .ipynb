```python
contacts = {
    'Chiranjay': '+1-519-574-4155',
    'Alice': '+1-519-574-4156'
}
def add_contact(name, phone):
    contacts[name] = phone
add_contact('CJ', 2)
def search_contact(name):
    if name in contacts:
        print(contacts[name])
    else:
        print(f"Sorry, {name} doesn't exist in phonebook")
search_contact('Alice')
def list_all_contacts():
    for name, number in contacts.items():
        print (name, number)

list_all_contacts()
contacts.get('Chiranjay')

def delete_contact(name):
    if name in contacts:
        contacts.pop(name)
        print(f'Contact details of {name} has been deleted from the contact book')
    else:
        print(f"Sorry, {name} doesn't exist in phonebook")

delete_contact('Chiranjay')      
```
