```python
add = lambda x, y: x + y
print(add(3, 4))  # 7
```

```python
temps = [0, 20, 37, 100]
converted = []
for t in temps:
    converted.append((t * 9/5) + 32)
print(converted)
```

```python
temps = [0, 20, 37, 100]
converted = list(map(lambda t: (t * 9/5) + 32, temps))
print(converted)
```

```python
number_list = [1,2,3,4,5,6,7]
for n in number_list:
    if n % 2 != 0:
        print('Odd')
    else:
        print('Even')
```

```python
even_or_odd = lambda x: 'Even' if x % 2 == 0 else 'Odd'
even_or_odd(7)
```

```python
names = ['ALICE', 'BOB', 'DIANA', 'TOM']
new_list = list(map(lambda names: names.lower(), names))
print(new_list)
```
