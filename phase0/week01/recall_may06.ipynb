```python
#A dictionary with 3 people, each with a name and age
phone_book = {
'CJ': 30,
'DJ': 26,
'EJ': 20
}
#Loop through it and print only people over 25
for key, value in phone_book.items():
    if value > 25:
        print(f'Name is {key} and age is {value}')
```
