```python
# Leverage range
for i in range(1,11):
    print(i)
```

```python
#Test loop
i = 1
while i < 6:
    print(i)
    i = i+1
```

```python
#Learn how to use break
for i in range(1,11):
    print(i)
    if i == 7:
        break
```

```python
#Add all numbers from 1 to 100
old_i = 0
for i in range(1,101):  
    
    old_i += i
print(old_i)
```

```python
#Print all even numbers
for i in range(1,51):
    if i % 2 == 0:
        print(i)
```

```python
#Multiplication table for 7. Print 7x1 through 7x10
table = 7
for i in range(1,11):
    result = table*i
    print(f'{table} * {i} = {result}')
```

```python
#Countdown from 20 to 0 in steps of 3
for i in range(20, -1, -3):
    print(i)
```
