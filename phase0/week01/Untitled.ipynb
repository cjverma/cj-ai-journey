```python
word_list = ['dam', 'tango', 'orange']
def find_first_long_word(word_list):
    for i in word_list:
        if len(i) > 5:
            return i
    print("No long words")
find_first_long_word(word_list)
```

```python
numbers = [1, 2, 3, 4, 5]

squared_odds = []

for num in numbers:
    if num % 2 != 0:
        squared_odds.append(num ** 2)



squared_odds = [num**2 for num in numbers if num % 2!= 0]
print(squared_odds)
```
