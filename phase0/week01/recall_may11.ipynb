```python
#find_first_even
list_number = [1,5,3,4,5,6,7,8,9,9]
def find_first_even(list_number):
    for i in list_number:
        if i % 2 == 0:
            return i
    print("No even numbers")

find_first_even()
```

```python
#Write loop using enumerate().
languages = ["Python", "Java", "C++"]
for index, language in enumerate(languages):
    print(index, language)
```

```python
models = ["CNN", "Transformer", "XGBoost"]
scores = [0.91, 0.97, 0.84]
for model, score in zip(models, scores):
    if score > 0.90:
        print( model, score)
```

```python
words = ["AI", "python", "ML", "data"]

long_words = []

for word in words:
    if len(word) > 2:
        long_words.append(word)

long_words = [word for word in words if len(word) > 2]
print(long_words)
long_wordss = [word for word in words if word.upper == word]
print(long_wordss)
```
