```python
#Requirements: takes 2 parameters:name and role and prints:My name is Alice and I am a ML Engineer

def introduce(name, role):
    print(f'My name is {name} and I am a {role}')
introduce('Alice', 'ML Engineer')
```

```python
def multiply(a,b):
    return a*b

result = multiply(2,3)
print(result)
```

```python
#find_first_negative
numbers = [3, 5, -2, -9]
def find_first_negative(numbers):
    for i in numbers:
        if i < 0:
            return i
    return "No Negative"
find_first_negative(numbers)
```

```python
# a function that only returns number of full batches only. 
def calculate_batches(dataset_size, batch_size):
    return dataset_size // batch_size

score = calculate_batches(1000,128)
print(score)
```

```python
#Manual counter for items in list
fruits = ["apple", "banana", "orange"]
i = 0
for fruit in fruits:
    print(i, fruit)
    i = i+1
```

```python
# Testing enumeration
models = ["Linear Regression", "Random Forest", "XGBoost"]
for index, model in enumerate(models):
    print(index, model)
```

```python
#Testing Zip
students = ["Alice", "Bob", "Charlie"]
grades = [90, 85, 95]
rankings = [1,2,3]
for student, grade, ranking in zip(students, grades, rankings):
    print(student, grade, ranking)
```

```python
# Print ONLY models with accuracy greater than 0.90.
models = ["CNN", "Transformer", "Random Forest"]
accuracies = [0.91, 0.97, 0.85]
for model, accuracy in zip(models, accuracies):
    if accuracy > 0.90:
        print(model, accuracy)
```
