```python
def find_maximum(*args):
    largest= args[0]
    for i in args:
        if i > largest:
            largest = i
    return largest
find_maximum(1,2,7,4,5)
```

```python
def count_even_numbers(*args):
    x = 0
    for i in args:
        if i % 2 == 0:
            x = x + 1
    print(x)
count_even_numbers(1,2,3,4,5,6,7,8)
```

```python
def find_shortest_word(*args):
    x = args[0]
    for i in args:
        if len(i) < len(x):
            x = i
    return x
            
find_shortest_word('a', 'bb', 'ccc')
```

```python
def print_all_uppercase(*args):
    for i in args:
        print(i.upper())
print_all_uppercase("ml", "python", "ai")
```

```python
def contains_negative(*args):
    for i in args:
        if i < 0:
            return True
    return False
contains_negative(1,4,-3, 5)
```

```python
def print_keys(**kwargs):
    for i in kwargs.keys():
        print(i)
print_keys(name= 'CJ', role= 'Engineer', age= 3)
```

```python
def print_string_values(**kwargs):
    for i in kwargs.values():
        if type(i) == str:
            print (i)
print_string_values(name="Alice", age=25, city="Toronto")
```

```python
def count_numeric_values(**kwargs):
    x = 0
    for i in kwargs.values():
        if type(i) == int or type(i) == float:
            x = x+1
    return x
count_numeric_values(name="Alice", age=25, score=91.5)
```
