```python
numbers = [1,2,3,4,5]
def find_average(numbers):
    total = 0
    for i in numbers:
        total = total + i
    return total / int(len(numbers))
find_average(numbers)
```

```python
person_details = {
    "name": "Alice",
    "age": 25,
    "score": 91.5
}
def print_numeric_values(person_details):
    for values in person_details.values():
        if type(values) == int or type(values) == float:
                print(values)
print_numeric_values(person_details)
```

```python
numbers = [1, 2, 3, 4, 5]
result = [num ** 2 for num in numbers if num % 2 == 0]
print(result)
```

```python
make_positive = lambda x: x * -1 if x < 1 else x

make_positive(0)
```

```python
old_list = list(filter(lambda x : x % 2 != 0, [1,2,3,4,5]))
print(old_list)
```

```python
def find_smallest(*args):
    smallest = args[0]
    for i in args:
        if i < smallest:
            smallest = i
    return smallest
find_smallest(1,2,3,4,5,2,3, -1)
```

```python
def print_long_string_values(**kwargs):
    for i in kwargs.values():
        if type(i) is str and len(i) > 4:
            print(i)
print_long_string_values(
    name="Alice",
    city="Toronto",
    age=25
)
```

```python
numbers = [1, 2, 3, 4]

result = list(
    filter(
        lambda x: x % 2 == 0,
        map(lambda x: x * 3, numbers)
    )
)

print(result)
```
