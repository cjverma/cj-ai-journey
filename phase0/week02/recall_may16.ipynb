```python
numbers = [5, "hello", 3]
def find_second_largest(numbers):
    unique = list(set(numbers))
    if len(unique) < 2:
        raise ValueError ("Your iterable contains less than 1 value")
    if unique[0] > unique[1]:
        largest = unique[0]
        second_largest = unique[1]
    else:
        largest = unique[1]
        second_largest = unique[0]

    for num in unique[2:]:
        if num > largest and type(num) == int or float:
            second_largest = largest
            largest = num

        elif num > second_largest and num != largest and type(num) == int or float:
            second_largest = num

    return second_largest
find_second_largest(numbers)
```

```python
def count_positive(*args):

    total = 0

    for num in args:
        if num > 0:
            total = total + num

    return total

print(count_positive(-2, 3, 5, -1, 4))
```

```python
def count_positive(*args):

    total = 0

    for num in args:
        if num > -1:
            total = total + 1

    return total

print(count_positive(-2, 3, 5, -1, 4))
```

```python
users = {
    "alice": {
        "password": "ml123",
        "locked": False,
        "attempts": 1
    }
}
def authenticate(username, password):
    for i, p in users.items():
        if i != username:
            print('User not found')
        if i == username and p['password'] == password:
            print("Successful Login")
authenticate('orange', 'ml123')
```

```python
prices = {
    "apple": 2,
    "banana": 1,
    "mango": 5,
    "orange": 3
}
expensive = {
    fruit : price
    for fruit, price in prices.items()
    if price >= 3
}
print(expensive.items())
```

```python
numbers = [1, 2, 3, 4]

squares = {
    num : num ** 2
    for num in numbers
}
print(squares)
```

```python
words = ["ml", "python", "ai"]

lengths = {
    word : len(word)
    for word in words
}

print(lengths)
```

```python
numbers = [1, 2, 3, 4, 5]

even_squares = {
    num : num ** 2
    for num in numbers
    if num % 2 == 0
}

print(even_squares)
```

```python
names = ["alice", "bob", "charlie"]

uppercase_names = {
    name : name.upper()
    for name in names
}
print(uppercase_names)
```
